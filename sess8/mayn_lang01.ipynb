```python
from typing import TypedDict
from langgraph.graph import StateGraph, START , END
from IPython.display import Image, display

class MYState(TypedDict):
    message: str

#Node => Fucntion
#state["message"] => READES THE STATE AND VALIDTE WITH YOU STATE 
def greeting(state: MYState):
    return {
        "message": state["message"] + "welcome to our class"
    }

def greetingGame(state:MYState):
    return {
          "message": state["message"] + "welcome Games"
    }

#Creating the graph"""  """
builder = StateGraph(MYState)

#add nodes
builder.add_node("greeting",greeting)
builder.add_node("greetingGame",greetingGame)

#connect graph
builder.add_edge(START,"greeting")
builder.add_edge("greeting","greetingGame")
builder.add_edge("greetingGame",END)

#Validate & Executable langgraph
app = builder.compile()
result=app.invoke({"message":"Hello "})
display(Image(app.get_graph().draw_mermaid_png()))
print(result)
```
