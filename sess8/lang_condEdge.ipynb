```python
from typing import TypedDict
from langgraph.graph import StateGraph, START , END
from IPython.display import Image, display

#Ticket Booking agent
class State(TypedDict):
    destination: str
    available: bool
    message:str

#node
def check_ticket(state: State):
    print("checking ticker avalibility...")
    available= state["destination"].lower() == "mumbai"
    return {"available":available}

def book_ticket(state:State):
    return {"message":f"Ticket booked for {state["destination"]}"}

def show_unavailable(state:State):
    return {"message":f"no tickers available for {state["destination"]}"}

def route_booking(state:State):
    if state["available"]:
         return "booking_done"
    
    return "unavailable_ticket"

builder = StateGraph(State)

builder.add_node("check",check_ticket)
builder.add_node("booking_available",book_ticket)
builder.add_node("unavailable",show_unavailable)

builder.add_edge(START,"check")
builder.add_conditional_edges(
    "check",
    route_booking,
    {
        "booking_done":"booking_available",
        "unavailable_ticket":"unavailable"
    }
)

builder.add_edge("booking_available",END)
builder.add_edge("unavailable",END)

app=builder.compile()

result=app.invoke({
    "destination":"Mumbai",
    "available":False,
    "message":""
})

print(result["message"])

display(Image(app.get_graph().draw_mermaid_png()))

```
